# 2. Why Do We Need Version Control?

Imagine you are working on a Python project.

You have:

```text
project.py
```

You make changes.

Then you realize that the new version has errors.

You might start creating:

```text
project_final.py
project_final_v2.py
project_final_v3.py
project_final_latest.py
project_final_latest_2.py
```

This quickly becomes difficult to manage.

### Version control solves this problem.

Instead of manually creating multiple copies, Git keeps track of changes.

```text
Version 1
   ↓
Version 2
   ↓
Version 3
   ↓
Version 4
```

You can inspect what changed and return to earlier versions when necessary.

---

# 3. What is Version Control?

**Version Control System (VCS)** is a system that records changes to files over time.

It allows developers to:

* Track changes
* Compare versions
* Restore previous versions
* Work with other developers
* Maintain project history
* Create separate development branches

### Examples of Version Control Systems

* Git
* Subversion (SVN)
* Mercurial

Today, **Git** is one of the most widely used version control systems.

---

# 4. What is Git?

**Git is a distributed version control system.**

It was designed to track changes in source code and other files.

Git can be used locally on your computer without GitHub.

For example:

```text
Your Computer
     │
     └── Git Repository
             │
             ├── project.py
             ├── data.csv
             └── README.md
```

Git tracks the history of changes to these files.

---

# 5. What is GitHub?

**GitHub is an online platform for hosting Git repositories and collaborating on software projects.**

Think of:

```text
Git
↓
Version Control System

GitHub
↓
Online platform built around Git
```

Git can work without GitHub.

GitHub uses Git to provide additional features such as:

* Remote repositories
* Collaboration
* Pull requests
* Code review
* Issue tracking
* Project management
* Team access

---

# 6. Git vs GitHub

This is one of the most important concepts for beginners.

| Git                     | GitHub                                            |
| ----------------------- | ------------------------------------------------- |
| Software                | Online platform/service                           |
| Version control system  | Git repository hosting and collaboration platform |
| Works locally           | Primarily used online                             |
| Tracks changes          | Hosts and collaborates around repositories        |
| Command line/tool       | Web interface + Git integration                   |
| Can work without GitHub | Uses Git repositories                             |

### Simple analogy

Think about:

```text
Git = notebook that records versions

GitHub = online place where you store and share that notebook
```

# 8. Git Configuration

Before making commits, configure your identity.

```bash
git config --global user.name "Your Name"
```

Then:

```bash
git config --global user.email "your@email.com"
```

Check configuration:

```bash
git config --global --list
```

### Why?

Git uses this information to associate commits with an author.

# 9. Git Repository

A **repository**, or repo, is a project managed by Git.

Example:

```text
data-science-project/
│
├── analysis.py
├── dataset.csv
├── README.md
└── requirements.txt
```

When Git is initialized:

```bash
git init
```

Git creates a hidden `.git` directory.

```text
data-science-project/
│
├── .git/
├── analysis.py
├── dataset.csv
└── README.md
```

The `.git` directory contains Git's repository metadata and history.

**Do not manually modify or delete `.git`.**

# 10. Basic Git Workflow

The fundamental Git workflow is:

```text
Working Directory
       ↓
     git add
       ↓
Staging Area
       ↓
   git commit
       ↓
Local Repository
       ↓
    git push
       ↓
GitHub Remote Repository
```

And when other people make changes:

```text
GitHub
   ↓
git pull
   ↓
Local Repository
```

---

# 11. Three Important Areas

Git beginners should understand these three concepts.

## Working Directory

The files you are currently working on.

```text
project.py
README.md
data.csv
```

## Staging Area

Files selected for the next commit.

Command:

```bash
git add
```

## Repository

The committed history.

Command:

```bash
git commit
```

# 13. `git status`

One of the most useful commands.

```bash
git status
```

It tells you:

* Current branch
* Modified files
* Untracked files
* Staged files
* Files ready to commit

Students should get into the habit of using:

```bash
git status
```

frequently.

---

# 14. `git add`

Suppose:

```text
analysis.py
README.md
```

have changed.

To stage one file:

```bash
git add analysis.py
```

To stage multiple files:

```bash
git add analysis.py README.md
```

To stage all appropriate changes in the current directory:

```bash
git add .
```

---

# 15. What Does "Staging" Mean?

Think of staging as preparing a package.

```text
Working Directory
      │
      │ git add
      ↓
Staging Area
      │
      │ git commit
      ↓
Repository
```

You decide which changes should be included in the next snapshot.

---

# 16. `git commit`

A commit records staged changes in Git's history.

```bash
git commit -m "Add initial data analysis"
```

A good commit message describes **what the commit changes**.

Good:

```text
Add data cleaning function
```

Less useful:

```text
changes
```

or:

```text
update
```

---

# 17. Commit as a Snapshot

Think of a commit as a checkpoint.

```text
Commit 1
   ↓
Initial project

Commit 2
   ↓
Add data cleaning

Commit 3
   ↓
Add visualization

Commit 4
   ↓
Fix missing values
```

This creates a project history.

---

# 18. Viewing Commit History


A shorter version:

Example:

```text

c71d9f2 Add data cleaning
```

# 21. Connecting Local Repository to GitHub

After creating a GitHub repository, Git can be connected to it with a remote.

A common command is:

```bash
git remote add origin <repository-url>
```

Check the remote:

```bash
git remote -v
```

`origin` is a conventional name for the primary remote repository.

---

# 22. `git push`

`git push` sends local commits to a remote repository.

Basic concept:

```text
Local Repository
       ↓
   git push
       ↓
GitHub Repository
```

Example:

```bash
git push
```

For the first push on a new branch, you may commonly see:

```bash
git push -u origin main
```

The `-u` option establishes an upstream relationship, making future pushes simpler.

---

# 23. `git pull`

`git pull` brings changes from a remote repository into your local working environment.

```bash
git pull
```

Conceptually:

```text
GitHub
   ↓
git pull
   ↓
Local Repository
```

This is especially important when working with teammates.

---

# 24. `git clone`

Suppose a project already exists on GitHub.

Instead of manually downloading individual files, use:

```bash
git clone <repository-url>
```

Conceptually:

```text
GitHub Repository
       ↓
   git clone
       ↓
Your Computer
```

After cloning:

```bash
cd repository-name
```

Then:

```bash
git status
```

---

# 25. `clone` vs `pull`

Beginners often confuse these.

### `git clone`

Used to get a repository for the first time.

```text
GitHub → Your Computer
```

### `git pull`

Used to update an existing local repository.

```text
GitHub → Existing Local Repository
```

# 28. Branches

A **branch** allows developers to work on a separate line of development.

Suppose the main project is:

```text
main
 │
 ├── Version 1
 ├── Version 2
 └── Version 3
```

You want to experiment with a new feature.

Create:

```text
feature-analysis
```

Now:

```text
main
 │
 ├── Version 1
 ├── Version 2
 └── Version 3
          \
           feature-analysis
```

The feature can be developed separately without immediately changing `main`.

---

# 29. Why Use Branches?

Branches are useful for:

* New features
* Bug fixes
* Experiments
* Team development
* Testing changes

For example:

```text
main
├── feature-dashboard
├── feature-data-cleaning
└── bugfix-missing-values
```

# 33. `git merge`

Suppose:

```text
main
   \
    feature-analysis
```

You finish the feature.

You can merge the feature into `main`.

First switch to main:

```bash
git switch main
```

Then:

```bash
git merge feature-analysis
```

Conceptually:

```text
feature-analysis
       │
       │
       └──────→ main
```

---

# 34. Merge Conflicts

Sometimes two people change the same part of the same file.

For example:

### Developer A

```python
model = "linear_regression"
```

### Developer B

```python
model = "random_forest"
```

Git may not know which version should be kept.

This creates a **merge conflict**.

---

# 35. What Does a Conflict Look Like?

Git may mark the file with conflict markers:

```text
<<<<<<< HEAD
model = "linear_regression"
=======
model = "random_forest"
>>>>>>> feature-model
```

The developer must decide what the final code should be.

After resolving:

```bash
git add .
```

Then:

```bash
git commit
```

# 37. What is a Pull Request?

A **Pull Request (PR)** is a request to merge changes from one branch into another.

For example:

```text
feature-analysis
       ↓
Pull Request
       ↓
main
```

A PR allows teammates to:

* Review code
* Comment
* Suggest changes
* Discuss implementation
* Approve changes
* Merge changes

# 38. Important Terminology

| Term         | Meaning                                                   |
| ------------ | --------------------------------------------------------- |
| Repository   | Project managed by Git                                    |
| Commit       | Recorded snapshot of changes                              |
| Branch       | Separate line of development                              |
| Remote       | Repository hosted elsewhere                               |
| Clone        | Copy a remote repository locally                          |
| Push         | Send local commits to remote                              |
| Pull         | Bring remote changes into local repo                      |
| Merge        | Combine changes from branches                             |
| Pull Request | Request to merge changes on GitHub                        |
| Conflict     | Git cannot automatically determine how to combine changes |
